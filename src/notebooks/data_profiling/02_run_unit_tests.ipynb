```python
import os
import sys
import pytest

# Prevent Python from attempting to write .pyc files to the Workspace filesystem
os.environ["PYTHONDONTWRITEBYTECODE"] = "1"

# Resolve path to the repository root
current_dir = os.getcwd()
repo_root = os.path.abspath(os.path.join(current_dir, "../../../"))
test_file_path = os.path.join(repo_root, "tests", "01_dataprofiling.py")

# Ensure the repo root is on sys.path so modules can import cleanly
if repo_root not in sys.path:
    sys.path.insert(0, repo_root)

# Run pytest with cache and assertion-rewrite disabled for Workspace Files compatibility
exit_code = pytest.main([
    "-v",
    "-p", "no:cacheprovider",
    "--assert=plain",
    test_file_path
])

assert exit_code == 0, f"Unit tests failed with exit code: {exit_code}"
```
